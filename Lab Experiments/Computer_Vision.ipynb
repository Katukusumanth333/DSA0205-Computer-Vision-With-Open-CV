```python
#Program 6
from google.colab import files
from IPython.display import clear_output
import cv2
import matplotlib.pyplot as plt
import time

# Upload Video
uploaded = files.upload()
video_name = list(uploaded.keys())[0]

cap = cv2.VideoCapture(video_name)

while cap.isOpened():
    ret, frame = cap.read()

    if not ret:
        break

    frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    clear_output(wait=True)
    plt.imshow(frame)
    plt.axis('off')
    plt.show()

    # Slow Motion
    time.sleep(0.1)

    # For Fast Motion use:
    # time.sleep(0.01)

cap.release()
```

```python
#program 8
# Scaling an Image (Bigger and Smaller)

from google.colab import files
import cv2
import matplotlib.pyplot as plt

# Upload Image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read Image
img = cv2.imread(filename)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Bigger Image (2x)
bigger = cv2.resize(img, None, fx=2, fy=2)

# Smaller Image (0.5x)
smaller = cv2.resize(img, None, fx=0.5, fy=0.5)

# Display Images
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(img)
plt.title("Original")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(bigger)
plt.title("Bigger")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(smaller)
plt.title("Smaller")
plt.axis('off')

plt.show()
```

```python
#program 10
from google.colab import files
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Upload Image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read Image
img = cv2.imread(filename)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Get Image Size
rows, cols = img.shape[:2]

# Translation Matrix
# Move 100 pixels right and 50 pixels down
M = np.float32([[1, 0, 100],
                [0, 1, 50]])

# Translate Image
moved = cv2.warpAffine(img, M, (cols, rows))

# Display Images
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(moved)
plt.title("Translated Image")
plt.axis('off')

plt.show()
```

```python
#program 11
from google.colab import files
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Upload Image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read Image
img = cv2.imread(filename)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

rows, cols = img.shape[:2]

# Affine Transformation Points
pts1 = np.float32([[50,50], [200,50], [50,200]])
pts2 = np.float32([[10,100], [200,50], [100,250]])

# Transformation Matrix
M = cv2.getAffineTransform(pts1, pts2)

# Apply Affine Transformation
affine = cv2.warpAffine(img, M, (cols, rows))

# Display Images
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(affine)
plt.title("Affine Transformed")
plt.axis('off')

plt.show()
```

```python
#program 12
from google.colab import files
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Upload Image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read Image
img = cv2.imread(filename)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

rows, cols = img.shape[:2]

# Source Points
pts1 = np.float32([[50,50],
                   [300,50],
                   [50,300],
                   [300,300]])

# Destination Points
pts2 = np.float32([[10,100],
                   [300,50],
                   [100,300],
                   [300,250]])

# Perspective Transformation Matrix
M = cv2.getPerspectiveTransform(pts1, pts2)

# Apply Perspective Transformation
perspective = cv2.warpPerspective(img, M, (cols, rows))

# Display Images
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(perspective)
plt.title("Perspective Transform")
plt.axis('off')

plt.show()
```

```python
#program 13
from google.colab import files
import cv2
import numpy as np
from IPython.display import Video

# Upload Video
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read Video
cap = cv2.VideoCapture(filename)

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

# Output Video
out = cv2.VideoWriter(
    'perspective_video.mp4',
    cv2.VideoWriter_fourcc(*'mp4v'),
    fps,
    (width, height)
)

# Source Points
pts1 = np.float32([
    [50,50],
    [width-50,50],
    [50,height-50],
    [width-50,height-50]
])

# Destination Points
pts2 = np.float32([
    [100,100],
    [width-100,50],
    [50,height-100],
    [width-100,height-50]
])

# Perspective Matrix
M = cv2.getPerspectiveTransform(pts1, pts2)

# Process Video
while True:
    ret, frame = cap.read()

    if not ret:
        break

    transformed = cv2.warpPerspective(
        frame,
        M,
        (width, height)
    )

    out.write(transformed)

cap.release()
out.release()

print("Perspective Transformed Video Saved")
```
