```
#Histogram Equalization
from google.colab import files
import cv2
import matplotlib.pyplot as plt

# Upload Image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read Image
img = cv2.imread(filename, 0)

# Histogram Equalization
enhanced = cv2.equalizeHist(img)

# Display Full Images
fig, ax = plt.subplots(1, 2, figsize=(5, 3))

ax[0].imshow(img, cmap='gray')
ax[0].set_title("Original Image")
ax[0].axis('off')

ax[1].imshow(enhanced, cmap='gray')
ax[1].set_title("Histogram Equalization")
ax[1].axis('off')

plt.tight_layout()
plt.show()
```

```
#Gaussian Blur (Noise Removal)
from google.colab import files
import cv2
import matplotlib.pyplot as plt

uploaded = files.upload()
filename = list(uploaded.keys())[0]

img = cv2.imread(filename)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

blur = cv2.GaussianBlur(img,(7,7),0)

fig, ax = plt.subplots(1,2,figsize=(5,3))

ax[0].imshow(img)
ax[0].set_title("Original Image")
ax[0].axis('off')

ax[1].imshow(blur)
ax[1].set_title("Gaussian Blur")
ax[1].axis('off')

plt.tight_layout()
plt.show()
```

```
#Binary Threshold Segmentation
from google.colab import files
import cv2
import matplotlib.pyplot as plt

uploaded = files.upload()
filename = list(uploaded.keys())[0]

img = cv2.imread(filename,0)

_, binary = cv2.threshold(img,127,255,cv2.THRESH_BINARY)

fig, ax = plt.subplots(1,2,figsize=(5,3))

ax[0].imshow(img,cmap='gray')
ax[0].set_title("Original Image")
ax[0].axis('off')

ax[1].imshow(binary,cmap='gray')
ax[1].set_title("Binary Threshold")
ax[1].axis('off')

plt.tight_layout()
plt.show()
```

```
#Image Smoothing
import cv2
from google.colab import files
import matplotlib.pyplot as plt

# Upload image
uploaded = files.upload()

# Read image
image_path = list(uploaded.keys())[0]
img = cv2.imread(image_path)

# Convert BGR to RGB
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Image Smoothing
smooth = cv2.blur(img, (5,5))

# Display
plt.figure(figsize=(6,6))
plt.imshow(smooth)
plt.axis("off")
plt.title("Image Smoothing")
plt.show()
```

```
#Image Sharpening
import cv2
import numpy as np
from google.colab import files
import matplotlib.pyplot as plt

uploaded = files.upload()

image_path = list(uploaded.keys())[0]
img = cv2.imread(image_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

kernel = np.array([[0,-1,0],
                   [-1,5,-1],
                   [0,-1,0]])

sharp = cv2.filter2D(img, -1, kernel)

plt.figure(figsize=(6,6))
plt.imshow(sharp)
plt.axis("off")
plt.title("Image Sharpening")
plt.show()
```
